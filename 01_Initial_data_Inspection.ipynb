**Data Understanding**

```python
# Check number of rows and columns
print("Rows and Columns:", df.shape)

# Show all column names
print("\nColumn Names:")
print(df.columns.tolist())

# Check data types
print("\nData Types:")
print(df.dtypes)

# Check basic statistics
print("\nBasic Statistics:")
display(df.describe(include="all"))
```

```python
# Check unique values in each column
for column in df.columns:
    print(f"\n{column}:")
    print("Unique values:", df[column].nunique())
    print(df[column].dropna().unique()[:10])
```

**Data Quality Check 🔍**

```python
# Check missing values
missing_values = df.isnull().sum()

print("Missing Values:")
print(missing_values[missing_values > 0])
```

```python
# Check duplicate rows
duplicate_count = df.duplicated().sum()

print("Number of duplicate rows:", duplicate_count)
```

```python
# Check categorical columns
print("Categorical Columns:")
print(df.select_dtypes(include="object").columns.tolist())
```

```python
# Check columns containing 'date'
date_columns = [col for col in df.columns if 'date' in col.lower()]

print("Possible Date Columns:")
print(date_columns)
```

**Data Quality Report**

```python
# Create a simple data quality report

quality_report = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Missing_Values": df.isnull().sum().values,
    "Unique_Values": df.nunique().values
})

quality_report
```

```python
# Check percentage of missing values

quality_report["Missing_Percentage"] = (
    quality_report["Missing_Values"] / len(df) * 100
).round(2)

quality_report
```

**Check unusual values**

```python
# Check negative values in numerical columns

numeric_columns = df.select_dtypes(include="number").columns

for column in numeric_columns:
    negative_count = (df[column] < 0).sum()
    if negative_count > 0:
        print(column, "->", negative_count, "negative values")
```

```python
# Data Cleaning

# Convert date to datetime
df["Order_Date"] = pd.to_datetime(df["Order_Date"])

# Remove extra spaces from text columns
text_columns = df.select_dtypes(include="object").columns

for column in text_columns:
    df[column] = df[column].str.strip()

# Create Revenue column
df["Revenue"] = df["Quantity"] * df["Unit_Price"]

# Final checks
print("Dataset Shape:", df.shape)
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())

df.head()
```

```python
# EDA: Overall Sales Performance

print("Total Revenue:", df["Revenue"].sum())

print("Total Orders:", df["Order_ID"].nunique())

print("Total Customers:", df["Customer_Name"].nunique())

print("Total Quantity Sold:", df["Quantity"].sum())

print("Average Order Revenue:",
      df.groupby("Order_ID")["Revenue"].sum().mean())
```

```python
# EDA 1: Revenue by Product Category

category_revenue = (
    df.groupby("Product_Category")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

print(category_revenue)
```

```python
# EDA 2: Revenue by Country

country_revenue = (
    df.groupby("Country")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

print(country_revenue)
```

```python
# EDA 3: Revenue by Customer Segment

segment_revenue = (
    df.groupby("Customer_Segment")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

print(segment_revenue)
```

```python
# EDA 4: Monthly Revenue Trend

monthly_revenue = (
    df.groupby(df["Order_Date"].dt.to_period("M"))["Revenue"]
    .sum()
)

print(monthly_revenue)
```

```python
# EDA 5: Top Products by Revenue

product_revenue = (
    df.groupby("Product_Name")["Revenue"]
    .sum()
    .sort_values(ascending=False)
)

print("Top 10 Products by Revenue:")
print(product_revenue.head(10))
```

```python
import matplotlib.pyplot as plt

category_revenue.sort_values().plot(
    kind="barh",
    figsize=(8, 5)
)

plt.title("Revenue by Product Category")
plt.xlabel("Revenue")
plt.ylabel("Product Category")
plt.tight_layout()
plt.show()
```

```python
# Top 10 Products Revenue Chart

top_products = product_revenue.head(10).sort_values()

top_products.plot(
    kind="barh",
    figsize=(9, 6)
)

plt.title("Top 10 Products by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Product Name")
plt.tight_layout()
plt.show()
```

```python
# Monthly Revenue Trend

monthly_revenue.plot(
    kind="line",
    figsize=(12, 5),
    marker="o"
)

plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```


**SQLite import**


```python
import sqlite3
import pandas as pd
```

```python
df_sql = pd.read_csv("ecommerce_cleaned.csv")

print("Rows:", len(df_sql))
print("Columns:", len(df_sql.columns))

df_sql.head()
```

SQLite database **creation**

```python
conn = sqlite3.connect("ecommerce_sales.db")
```

**CSV data ko SQL table**

```python
df_sql.to_sql(
    "ecommerce_sales",
    conn,
    if_exists="replace",
    index=False
)

print("Table created successfully.")
```

```python
query = """
SELECT *
FROM ecommerce_sales
LIMIT 5;
"""

result = pd.read_sql_query(query, conn)

result
```

**Overall Sales Performance**

```python
query = """
SELECT
    SUM(Revenue) AS Total_Revenue,
    COUNT(DISTINCT Order_ID) AS Total_Orders,
    SUM(Quantity) AS Total_Quantity
FROM ecommerce_sales;
"""

result = pd.read_sql_query(query, conn)

result
```

**Revenue by Product Category**

```python
query = """
SELECT
    Product_Category,
    SUM(Revenue) AS Total_Revenue
FROM ecommerce_sales
GROUP BY Product_Category
ORDER BY Total_Revenue DESC;
"""

result = pd.read_sql_query(query, conn)

result
```

**Revenue by Country 🌍**

```python
query = """
SELECT
    Country,
    SUM(Revenue) AS Total_Revenue
FROM ecommerce_sales
GROUP BY Country
ORDER BY Total_Revenue DESC;
"""

result = pd.read_sql_query(query, conn)

result
```

**Revenue by Customer Segment**

```python
query = """
SELECT
    Customer_Segment,
    SUM(Revenue) AS Total_Revenue
FROM ecommerce_sales
GROUP BY Customer_Segment
ORDER BY Total_Revenue DESC;
"""

result = pd.read_sql_query(query, conn)

result
```

**Top 10 Products by Revenue**

```python
query = """
SELECT
    Product_Name,
    SUM(Revenue) AS Total_Revenue
FROM ecommerce_sales
GROUP BY Product_Name
ORDER BY Total_Revenue DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, conn)

result
```

**Monthly Revenue Trend 📈**

```python
query = """
SELECT
    strftime('%Y-%m', Order_Date) AS Month,
    SUM(Revenue) AS Total_Revenue
FROM ecommerce_sales
GROUP BY Month
ORDER BY Month;
"""

result = pd.read_sql_query(query, conn)

result
```
